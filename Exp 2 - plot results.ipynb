```python
import pandas as pd
import matplotlib.pyplot as plt
from itertools import cycle
```

```python
list_dimensions = ['num_objects', 'mag_unfair', 'num_hotspots', 'num_regions', 'num_stops']
for dim in list_dimensions :

    # Read CSV into a DataFrame
    df = pd.read_csv(f"./experiments/{dim}/data.csv")
    name_dimension = df.columns[1]


    # Keep the x-axis in the desired order
    grid_order = ["50", "75", "100", "200", "400", "600", "800", "1000", "All"]
    df["Subset of grids"] = df["Subset of grids"].astype(str)
    df["Subset of grids"] = pd.Categorical(df["Subset of grids"], categories=grid_order, ordered=True)


    metrics = ["Power", "Sensitivity", "PPV"]
    dim_values = sorted(df[name_dimension].unique())
    marker_cycle = cycle(['o', 's', '^', 'D', 'v', 'p', '*', 'X', '<', '>'])
    for metric in metrics:
        plt.figure(figsize=(8, 5))
        plt.rcParams.update({"font.size": 12,
                             "axes.titlesize": 16,
                             "axes.labelsize": 14,
                             "xtick.labelsize": 12,
                             "ytick.labelsize": 12,
                             "legend.fontsize": 11})
        
        for value in dim_values:
            subset = df[df[name_dimension] == value].sort_values("Subset of grids")
            plt.plot(
                subset["Subset of grids"],
                subset[metric],
                marker=next(marker_cycle),
                label=f"{value}"
            )
        
        plt.xlabel("Subset of grids")
        plt.ylabel(metric)
        plt.title(f"{name_dimension} - {metric} analysis")
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.ylim(0, 1)
        plt.tight_layout()
        # plt.show()
        plt.savefig(f"./experiments/exp_plots/{dim}_{metric}_analysis.pdf")
```
