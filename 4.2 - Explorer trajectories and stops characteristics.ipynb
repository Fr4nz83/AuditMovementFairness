```python
# Read the geopandas dataframe, from a parquet, containing the compressed synthetic trajectories.
path_stops = './data_simulator/huge_dataset/dataset_simulator_trajectories.compressed.parquet.stops.parquet'
columns_coords = ['lat', 'lng']
df_stops = pd.read_parquet(path_stops)
df_stops
```

```python
num_stops_per_uid = df_stops.groupby('uid').size()
min_time, max_time = df_stops['datetime'].min(), df_stops['leaving_datetime'].max()
span_time = max_time - min_time

print(min_time, max_time)
print(span_time)
num_stops_per_uid_per_day = num_stops_per_uid / 10
print(num_stops_per_uid_per_day)
print(num_stops_per_uid_per_day.median(), num_stops_per_uid_per_day.std())
print(df_stops['duration_secs'].median() / 60, df_stops['duration_secs'].std() / 60)
```
