```python
import pandas as pd
import geopandas as gpd
import numpy as np
```

```python
# Read the geopandas dataframe, from a parquet, containing the compressed synthetic trajectories.
path_dataset = './data_simulator/huge_dataset/dataset_simulator_trajectories.parquet'
df = pd.read_parquet(path_dataset)
```

```python
# Plot a heatmap of the trajectory points with Folium.
import folium
from folium.plugins import HeatMap

# HeatMap expects [lat, lng]
df_sample = df.sample(1000000)
heat_data = df_sample[['lat', 'lon']].values.tolist()

# center map on the average point
m = folium.Map(
    location=[df_sample['lat'].mean(), df_sample['lon'].mean()],
    zoom_start=13,
    prefer_canvas=True
)

HeatMap(heat_data, radius=2, blur=1, min_opacity=0.6).add_to(m)

m
```

```python
# Read the geopandas dataframe, from a parquet, containing the compressed synthetic trajectories.
path_stops = './data_simulator/huge_dataset/dataset_simulator_trajectories.compressed.parquet.stops.parquet'
df_stops = pd.read_parquet(path_stops)
df_stops


# Plot a heatmap of the trajectory points with Folium.
import folium
from folium.plugins import HeatMap

# HeatMap expects [lat, lng]
heat_data = df_stops[['lat', 'lng']].values.tolist()

# center map on the average point
m = folium.Map(
    location=[df_stops['lat'].mean(), df_stops['lng'].mean()],
    zoom_start=13,
    prefer_canvas=True
)

HeatMap(heat_data).add_to(m)

m
```
